## Aanmaken database 
### Maak een in-memory database aan met de gegevens uit de Excel bestanden.

De methode `myconnection()` is nodig om de connectie ook vanuit SQL te kunnen benaderen.

```python
import sqlite3
import pandas as pd

# maak een in-memory database
conn = sqlite3.connect(":memory:")

# methode die verwijst naar de in-memory database
def myconnection():
    return conn
```

### Lees de excelbestanden uit het pad en maak daar tabellen van in de database.

```python
# PAS HET PAD AAN NAAR DE LOCATIE WAAR JIJ DE DATA HEBT GEPLAATST
padnaam = "../../R/Datasets/De Rode Schoentjes"

# Maak de tabel met de budgetten aan
df = pd.read_excel(f"{padnaam}/Budget.xlsx", sheet_name="Budget")
df.to_sql("Budget", conn, if_exists="replace", index=False)

# Maak de tabel met de categorieën aan
df = pd.read_excel(f"{padnaam}/Categorie.xlsx", sheet_name="Categorie")
df.to_sql("Categorie", conn, if_exists="replace", index=False)

# Maak de tabel met de filialen aan
df = pd.read_excel(f"{padnaam}/Filiaal.xlsx", sheet_name="Filiaal")
df.to_sql("Filiaal", conn, if_exists="replace", index=False)

# Maak de tabel met de kassa's aan
df = pd.read_excel(f"{padnaam}/Kassa.xlsx", sheet_name="Kassa")
df.to_sql("Kassa", conn, if_exists="replace", index=False)

print(f"In-memory database is aangemaakt")
```

### Maak een tabel aan met datums
Leg per datum algemene gegevens vast als weekdag, maandnaam, enz. Dat is makkelijk om de gegevens te groeperen, maar zo kun je ook bepalen op welke dagen er géén omzet was.

```python
# Maak een lijst met alle datums in het jaar 2025 met hele weken in 2024 en 2026
datums = pd.date_range(start="2024-12-30", end="2026-01-04", freq="D")

# Genereer een datumtabel voor de gegevens per week, voor elke woensdag, etc.
iso = datums.isocalendar()
datumtabel = pd.DataFrame({
    "Datum": datums.strftime("%Y-%m-%d"),
    "Wdag": datums.dayofweek + 1,
    "Wd": datums.strftime("%a"),
    "WdNm": datums.strftime("%A"),
    "WkNr": iso.week,
    "JrWk": (
        iso.year.astype(str).str[-2:]
        + "/"
        + iso.week.astype(str).str.zfill(2)
    ),
    "MndNr": datums.month,
    "Mnd": datums.strftime("%b"),
    "MndNm": datums.strftime("%B"),
    "Mdag": datums.day,
    "Qrt": (
        datums.year.astype(str).str[-2:]
        + "Q"
        + datums.quarter.astype(str)
    ),
    "JrNr": datums.year,
    "Jdag": datums.dayofyear
})
datumtabel.to_sql("Datum", conn, if_exists="replace", index=False)
print("Tabel \"Datum\" is toegevoegd.") 
```
