```python
import pandas as pd
df = pd.read_excel('~/Documents/Marketing Campaign Dataset Excel.xlsx')
print(df.head())
```

```python
pd.set_option('display.max_columns', None)
pd.set_option('display.width', None)
pd.set_option('display.max_colwidth', None)
print(df.head())
```

```python
# CTR berekenen
df['CTR'] = (df['clicks'] / df['impressions'] * 100).round(2)

# Vervang inf waarden met 0
df['CTR'] = df['CTR'].replace([float('inf'), float('-inf')], 0)

# Toon eerste 10 rijen
print(df[['Date', 'campaign_type', 'impressions', 'clicks', 'CTR']].head(10))
```

```python
# CEL 4: CPC berekenen (Cost Per Click)
df['CPC'] = (df['spends'] / df['clicks']).round(2)
df['CPC'] = df['CPC'].replace([float('inf'), float('-inf')], 0)
print('CPC berekend. Eerste 10 rijen:')
print(df[['Date', 'campaign_type', 'spends', 'clicks', 'CPC']].head(10))
```

```python
# CEL 5: CPM berekenen (Cost Per 1000 Impressions)
df['CPM'] = (df['spends'] / df['impressions'] * 1000).round(2)
df['CPM'] = df['CPM'].replace([float('inf'), float('-inf')], 0)
print('CPM berekend. Eerste 10 rijen:')
print(df[['Date', 'campaign_type', 'spends', 'impressions', 'CPM']].head(10))
```

```python
# CEL 6: TOTALE KPI's berekenen
totale_clicks = df['clicks'].sum()
totale_impressions = df['impressions'].sum()
totale_spends = df['spends'].sum()
totale_ctr = (totale_clicks / totale_impressions * 100)
totale_cpc = (totale_spends / totale_clicks)
totale_cpm = (totale_spends / totale_impressions * 1000)

print('TOTALE METRIEK:')
print(f'Totale Spend: €{totale_spends:,.2f}')
print(f'Totale Impressions: {totale_impressions:,}')
print(f'Totale Clicks: {totale_clicks:,}')
print(f'\nTotale CTR: {totale_ctr:.2f}%')
print(f'Totale CPC: €{totale_cpc:.2f}')
print(f'Totale CPM: €{totale_cpm:.2f}')
```

```python
# CEL 7: KPI's PER CAMPAIGN TYPE
campaign_stats = df.groupby('campaign_type').agg({
    'spends': 'sum',
    'impressions': 'sum',
    'clicks': 'sum'
}).round(2)

campaign_stats['CTR_%'] = (campaign_stats['clicks'] / campaign_stats['impressions'] * 100).round(2)
campaign_stats['CPC_€'] = (campaign_stats['spends'] / campaign_stats['clicks']).round(2)
campaign_stats['CPM_€'] = (campaign_stats['spends'] / campaign_stats['impressions'] * 1000).round(2)

print('CAMPAGNE PERFORMANCE:')
print(campaign_stats)
```

```python
# CEL 8: Vergelijken met BENCHMARKS
benchmarks = {
    'Lead-Gen': {'CTR': 6.64, 'CPC': 5.42, 'CPM': 15},
    'Ecommerce': {'CTR': 3.27, 'CPC': 6.50, 'CPM': 4.85}
}

print('='*70)
print('VERGELIJKING MET NEDERLANDSE B2B BENCHMARKS')
print('='*70)
print(f'\nOnze Data (totaal):')
print(f'  CTR: {totale_ctr:.2f}%')
print(f'  CPC: €{totale_cpc:.2f}')
print(f'  CPM: €{totale_cpm:.2f}')

print(f'\nBenchmarks:')
print(f'  Lead-Gen    -> CTR: 6.64%, CPC: €5.42, CPM: €15')
print(f'  Ecommerce   -> CTR: 3.27%, CPC: €6.50, CPM: €4.85')

print(f'\nAFWIJKINGEN (vs Ecommerce benchmark):')
cpc_delta = ((totale_cpc - 6.50) / 6.50 * 100)
cpm_delta = ((totale_cpm - 4.85) / 4.85 * 100)
ctr_delta = ((totale_ctr - 3.27) / 3.27 * 100)

print(f'  CPC: {cpc_delta:+.1f}% (€{totale_cpc:.2f} vs €6.50)')
print(f'  CPM: {cpm_delta:+.0f}% (€{totale_cpm:.2f} vs €4.85)')
print(f'  CTR: {ctr_delta:+.1f}% ({totale_ctr:.2f}% vs 3.27%)')

print(f'\nCONCLUSIE:')
print(f'Problemen zijn met data BEWEZEN - niet zomaar aannames!')
print(f'Je betaalt significant meer per click en per 1000 impressies.')
```
