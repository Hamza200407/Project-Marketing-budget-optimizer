```python
woord = woorden[43].upper();
geraden = ["P", "O", "N"];
fouten = ["A", "E", "S", "D", "K"];

# voeg de letter "L" toe aan de letters die niet in het woord voorkomen.
fouten.append("L");
```

```python
# voorbereiding voor de uitvoer 
fig, ax = plt.subplots();
ax.set_xlim(-4,4);
ax.set_ylim(-1,7);
ax.set_aspect("equal");
ax.axis("off");

#galg staander
ax.plot([-2.5,-2.5],[-1,6])

# galgpoot links
ax.plot([-2.5,-3.5],[0.5,-1])

# galgpoot rechts
ax.plot([-2.5,-1.5],[0.5,-1])

# galg hoek
ax.plot([-2.5,-1.0],[4.5,6])

# galg ligger
ax.plot([-2.5,0],[6,6])

# galg touw
ax.plot([0,0],[6,4.8])

# hoofd
hoofd = plt.Circle((0,4), 0.7, fill=False)
ax.add_patch(hoofd);

# lijf
ax.plot([0,0], [3.3,1.5])

# linkerarm
ax.plot([-1,0], [2.8,2.8])

# rechterarm
ax.plot([0,1], [2.8,2.8])

# been links
ax.plot([0,-0.8],[1.5,0])

#been rechts
ax.plot([0,0.8],[1.5,0])

weergave = ""
for letter in woord:
    if letter in geraden:
        weergave += letter + " "
    else:
        weergave += "_ "
ax.text(3, 4, weergave, fontsize=48)
ax.text(3, 2, "Fouten: " + ", ".join(fouten), fontsize=15)

plt.show()
```

```python
letter = input("Gok een letter: ").upper().strip()

if not letter:
    print("Voer alsjeblieft een letter in!")
elif len(letter) > 1:
    print("Voer alsjeblieft maar één letter in!")
elif letter in geraden or letter in fouten:
    print(f"Je hebt {letter} al geprobeerd! Probeer een ander.")
else:
    if letter in woord:
        geraden.append(letter)
        print(f"GOED! {letter} zit in het woord!")
    else:
        fouten.append(letter)
        print(f"FOUT! {letter} zit niet in het woord!")

# Teken galg
fig, ax = plt.subplots(figsize=(10, 8))
ax.set_xlim(-4, 4)
ax.set_ylim(-1, 7)
ax.set_aspect("equal")
ax.axis("off")

ax.plot([-2.5, -2.5], [-1, 6], 'k-', linewidth=2)
ax.plot([-2.5, -3.5], [0.5, -1], 'k-', linewidth=2)
ax.plot([-2.5, -1.5], [0.5, -1], 'k-', linewidth=2)
ax.plot([-2.5, -1.0], [4.5, 6], 'k-', linewidth=2)
ax.plot([-2.5, 0], [6, 6], 'k-', linewidth=2)
ax.plot([0, 0], [6, 4.8], 'k-', linewidth=1)

if len(fouten) >= 1:
    hoofd = plt.Circle((0, 4), 0.7, fill=False, edgecolor='k', linewidth=2)
    ax.add_patch(hoofd)
if len(fouten) >= 2:
    ax.plot([0, 0], [3.3, 1.5], 'k-', linewidth=2)
if len(fouten) >= 3:
    ax.plot([-1, 0], [2.8, 2.8], 'k-', linewidth=2)
if len(fouten) >= 4:
    ax.plot([0, 1], [2.8, 2.8], 'k-', linewidth=2)
if len(fouten) >= 5:
    ax.plot([0, -0.8], [1.5, 0], 'k-', linewidth=2)
if len(fouten) >= 6:
    ax.plot([0, 0.8], [1.5, 0], 'k-', linewidth=2)

weergave = ""
for l in woord:
    if l in geraden:
        weergave += l + " "
    else:
        weergave += "_ "

ax.text(1.5, 5, weergave, fontsize=48, family='monospace', fontweight='bold')
if fouten:
    ax.text(1.5, 3.5, "Fouten: " + ", ".join(sorted(fouten)), fontsize=14)

plt.tight_layout()
plt.show()

if all(l in geraden for l in woord):
    print("JE HEBT GEWONNEN!")
    print(f"Het woord was: {woord}")

if len(fouten) >= 6:
    print("GAME OVER!")
    print(f"Het woord was: {woord}")
```
